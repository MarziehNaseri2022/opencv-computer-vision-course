```python
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
image=cv.imread("images/1.jpg")
plt.imshow(image[...,::-1])
```

```python
height,weidth= image.shape[:2]
cropped=image[50:370,300:500]
plt.imshow(cropped[...,::-1])
```

```python
cv.imshow("original",image)
cv.imshow("cropped",cropped)

cv.waitKey()
cv.destroyAllWindows()
```

```python
saved = cv.imwrite("./images/face_cropped.jpg", cropped)

if saved:
    print("saved")
else:
    print("didn't save.")
```
