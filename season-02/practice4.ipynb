```python
import cv2 as cv
import matplotlib.pyplot as plt 
import numpy as np
```

```python
img=np.full((500,500,3),(100,250,255),np.uint8)
plt.imshow(img[...,::-1])
```

```python
cv.rectangle(img,(150,200),(350,400),(0,50,100),-5)
cv.rectangle(img,(150,200),(350,400),(0,0,0),2)
# triangle
tgl=np.array([[100,200], [250,70], [400,200]],np.int32)

tgl = tgl.reshape((-1,1,2))

cv.polylines(img, [tgl], True, (0,0,0), 2)
cv.fillConvexPoly(img, tgl, (0, 0, 255))

```

```python
cv.rectangle(img,(220,300),(290,400),(0, 0, 0),2)
cv.rectangle(img,(220,300),(290,400),(0, 0, 255),-2)

cv.imshow("My beautifull Home", img)
plt.imshow(img[...,::-1]);plt.title("My beautifull Home")
cv.waitKey(0)
cv.destroyAllWindows()
```
