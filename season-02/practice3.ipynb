```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv.imread("images/1.jpg")
plt.imshow(img[...,::-1])
```

```python
img.shape
```

```python
cropped=img[15:395,192:572]
plt.imshow(cropped[...,::-1])
```

# PNG

```python
png_image = np.zeros((380,380,4), np.uint8)
png_image[:,:,0:3]=cropped
circle=np.zeros((380,380),np.uint8)
cv.circle(circle, (190, 190), 190, (255,255,255), -1) 
png_image[:,:,3]= circle
cv.imwrite("images/girl_circular.png", png_image)
```

## Transparent

```python
import cv2 as cv
import numpy as np


png_image2 = np.zeros((1140, 720, 4), dtype=np.uint8)
png_image2[:, :, 0:3] = img

alpha_mask = np.full((1140, 720), 76, dtype=np.uint8)

cv.circle(alpha_mask, (382, 205), 190, 255, -1)

png_image2[:, :, 3] = alpha_mask


cv.imwrite("images/girl_transparent.png", png_image2)
```
