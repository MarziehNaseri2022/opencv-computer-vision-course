```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
img= cv.imread("images/1.jpg")
plt.imshow(img[...,::-1])
```

```python
hsv= cv.cvtColor(img,cv.COLOR_BGR2HSV)

lower_blue = np.array([100, 50, 50])
upper_blue = np.array([130, 255, 255])

pink_l=np.array([140, 50, 50])
pink_h=np.array([170, 255, 255])

mask = cv.inRange(hsv, lower_blue, upper_blue)
mask2 = cv.inRange(hsv, pink_l, pink_h)

mask_inverted= cv.bitwise_not(mask)
mask2_inverted= cv.bitwise_not(mask2)

green_count = cv.countNonZero(mask2)
if green_count > 50:
    print(f"Yes here they are ! (count of pixels: {green_count})")
else:
    print("No,there aren't")

# ۶. showing mask
plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(img, cv.COLOR_BGR2RGB))
plt.title("Original")

plt.subplot(1, 2, 2)
plt.imshow(mask2_inverted, cmap='gray')
plt.title("Detected Mask")
plt.show()
```
